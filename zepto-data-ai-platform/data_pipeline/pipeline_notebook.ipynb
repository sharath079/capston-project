```python
# Cell 1: Imports
import pandas as pd
import sqlite3
from scrape_books import scrape_books
from clean_store import clean_books, store_sqlite

# Cell 2: Scrape data
df = scrape_books(pages=5)
df.head()

# Cell 3: Clean data
df_clean = clean_books(df)
df_clean.head()

# Cell 4: Store in SQLite
store_sqlite(df_clean)

# Cell 5: Run SQL queries
conn = sqlite3.connect("books.db")
query = "SELECT title, price_inr FROM books ORDER BY price_inr DESC LIMIT 5;"
pd.read_sql(query, conn)

# Cell 6: Demonstrate pd.merge equivalence
books = pd.read_sql("SELECT * FROM books", conn)
categories = pd.read_sql("SELECT * FROM categories", conn)
merged = pd.merge(books, categories, left_on="category_id", right_on="id")
merged.head()
```
